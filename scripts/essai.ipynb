```python
import math
import os
from glob import glob
from pyarrow.parquet import ParquetFile

OUTPUT_FILE = "../data/processed/"
#OUTPUT_FILE = "../data/raw/"

files = sorted(
    glob(os.path.join(OUTPUT_FILE,"nyc_taxi_cleaned.parquet"))
    #glob(os.path.join(OUTPUT_FILE,"*.parquet"))
)


total_fare_amount = 0
total_trip = 0
Min_amount = float("inf")
Max_amount = float("-inf")
for file in files :
    print(f"\n Traitement du Fichier {file}")

    parquet_file = ParquetFile(file)
    for batch in parquet_file.iter_batches(batch_size=100_000):
        df = batch.to_pandas()

        #Rechnen von Statistik_werte
        total_fare_amount += df["fare_amount"].sum()
        total_trip += len(df)       # df["fare_amount"].count()
        Min_amount = min(Min_amount, min(df["fare_amount"]))
        Max_amount = max(Max_amount, max(df["fare_amount"]))

mean = total_fare_amount/total_trip
print(f"\n\nSTATISTIK WERTE")
print(f"Total Fare amount = {total_fare_amount:.2f}")
print(f"Total trip = {total_trip}")
print(f"Mean = {mean:.2f}")
print(f"Max amount = {Max_amount:.2f}")
print(f"Min amount = {Min_amount:.2f}")
```
